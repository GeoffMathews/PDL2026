```python
# displays static in notebook
%matplotlib inline  
# displays interactive outside notebook
#%matplotlib   
# displays interactive in notebook
#%matplotlib notebook  

# to work correctly with numbers
import numpy as np

# plotting
import matplotlib.pyplot as plt
import matplotlib as mpl

# conversion between coordinate systems
import astropy.coordinates as coord
import astropy.units as u

#from astropy import units as u
from astropy.coordinates import SkyCoord

folder = ''

mpl.rcParams['font.family']='serif'
mpl.rcParams['font.size']=14
```

```python
# Units

M_earth = 5.97E24
r_earth = 6.37E6
T_earth = 254.0
M_sun = 1.989E30
L_sun = 3.839E26
r_sun = 6.957e8
AU = 1.496E11
pc = 3.086E16
sigma = 5.67E-8
G = 6.67E-11
```

# temperature vs distance

```python
rs = np.arange(0.1, 5.01, 0.01)
r = np.array([0.39, 0.72, 1.0, 1.52])
T_base = np.array([446, 328, 278, 226])
T_eff = np.array([440, 227, 255, 217])
T_surf = np.array([440, 740, 283, 222])

T_model = (1.0 * L_sun / (16 * np.pi * AU**2 * sigma) * (1.0 / rs**2))**(0.25)


fig = plt.figure(figsize=(16,8))
plt.plot(rs, T_model, 'black', linewidth=4)
plt.plot(r, T_base, 'ko', markersize=20, label='Baseline')
plt.plot(r,T_surf,'rs', label='Surface', markersize=15)
plt.plot(r,T_eff,'bD', label='Expected', markersize=10)
plt.plot([r[1]],[495],'r^', markersize=15)
plt.xlabel(r'Distance from sun [AU]', fontsize=20)
plt.ylabel(r'Temperature [Kelvin]', fontsize=20)
plt.tick_params(axis='both', which='major', labelsize=20)


#get handles and labels
handles, labels = plt.gca().get_legend_handles_labels()
#specify order of items in legend
order = [0,2,1]
#add legend to plot
plt.legend([handles[idx] for idx in order],[labels[idx] for idx in order], shadow=True, fontsize=20) 

#legend = plt.legend(loc='upper right', shadow=True, fontsize=20)
plt.xlim(0.1, 2.0)
plt.ylim(100,500)
plt.grid(linestyle="-", color='gray',linewidth=1.0, alpha=0.3)

x_lims = [0,100]
y_low = [273,273]
y_high = [373,373]
plt.fill_between(x_lims, y_low, y_high,color='gray',alpha=0.3)

plt.annotate('liquid water', xy=(1.3, 320),fontsize=20)
plt.annotate('Venus surface temperature: 740 K', xy=(0.75, 480),fontsize=20,color='gray')

planet_names = ['Mercury','Venus','Earth','Mars']
for label, x, y in zip(planet_names, r, T_eff):
    plt.annotate(label,xy=(x, y), xytext=(-20, -35), fontsize=20, textcoords='offset points', ha='center', va='bottom')
    #

plt.savefig("astrobio-solar-system-surface-temperatures-matchSim.jpg") #save as jpg
```
